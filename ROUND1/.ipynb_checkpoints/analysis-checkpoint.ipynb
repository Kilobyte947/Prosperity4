```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set(style='whitegrid', palette='muted', font_scale=1.1)
```

```python
price_files = {
    -2: "./data/prices_round_1_day_-2.csv",
    -1: "./data/prices_round_1_day_-1.csv",
     0: "./data/prices_round_1_day_0.csv"
}

trade_files = {
    -2: "./data/trades_round_1_day_-2.csv",
    -1: "./data/trades_round_1_day_-1.csv",
     0: "./data/trades_round_1_day_0.csv"
}
```

```python
price_day_minus1 = pd.read_csv('prices_round_1_day_-1.csv')
price_day_minus2 = pd.read_csv('prices_round_1_day_-2.csv')
```

```python
price_day_minus1.head()
```

```python
price_day_minus2.head()
```

```python
columns = price_day_minus1.columns[0].split(';')
#for i in range(len(columns)):
#    df_split = trades_day1.iloc[:,0].str.split(';',expand = True)
price_day_minus1_split = price_day_minus1.iloc[:,0].str.split(';',expand = True)
price_day_minus1_split.columns = columns
print(price_day_minus1_split.columns)
price_day_minus1_split.head()
```

```python
columns = price_day_minus2.columns[0].split(';')
#for i in range(len(columns)):
#    df_split = trades_day1.iloc[:,0].str.split(';',expand = True)
price_day_minus2_split = price_day_minus2.iloc[:,0].str.split(';',expand = True)
price_day_minus2_split.columns = columns
print(price_day_minus2_split.columns)
price_day_minus2_split.head()
```

```python
price_day_past_data = pd.concat([price_day_minus2_split, price_day_minus1_split], axis=0)
price_day_past_data.tail()
```

```python
price_day_past_data['day'] = price_day_past_data['day'].apply(pd.to_numeric)
price_day_past_data['timestamp'] = price_day_past_data['timestamp'].apply(pd.to_numeric)
price_day_past_data['day']*1000000
```

```python
price_day_past_data['time'] = price_day_past_data['timestamp'] + price_day_past_data['day']*1000000 + 2000000
pepper_root_data = price_day_past_data[price_day_past_data['product'] == 'INTARIAN_PEPPER_ROOT']
osmium_data = price_day_past_data[price_day_past_data['product'] == 'ASH_COATED_OSMIUM']
pepper_root_data.tail()
```

```python
pepper_root_data.head()
```

```python
pepper_root_data.set_index(pepper_root_data['time'], inplace=True)
pepper_root_data.head()
pepper_root_data['mid_price'] = pepper_root_data['mid_price'].apply(pd.to_numeric)
pepper_root_data.dtypes
```

```python
plt.plot(pepper_root_data['time'], pepper_root_data['mid_price'])
plt.show()
```
